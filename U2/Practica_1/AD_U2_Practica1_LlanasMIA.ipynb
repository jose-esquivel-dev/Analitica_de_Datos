```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
forbes_df = pd.read_csv('/content/drive/MyDrive/Análitica de Datos /Forbes_2000_top_company.csv')
forbes_df
```

5 de barras

```python
plt.figure(figsize=(15,9)) #Tamaño de la gráfica
sns.barplot(data=forbes_df.head(5), x='Organization Name', y='Total Employees', palette='rocket')
plt.title('Número de empleados por organización')

```

```python
plt.figure(figsize=(15,5)) #Tamaño de la gráfica
sns.barplot(data=forbes_df.head(7), x='Industry', y='Revenue (Billions)', palette='Set2', hue='Industry', errorbar=None, legend=False)
plt.title('Ingresos por Industria')
```

```python
plt.figure(figsize=(8,6)) #Tamaño de la gráfica
sns.barplot(data=forbes_df.tail(5), x='CEO', y='Profits (Billions)', palette='Paired', errorbar=None)
plt.title('Beneficios por CEO')
```

```python
plt.figure(figsize=(8,6)) #Tamaño de la gráfica
sns.barplot(data=forbes_df.tail(5), x='Year Founded', y='Assets (Billions)', palette='tab10', errorbar=None)
plt.title('Activos por año de Fundación')
```

```python
plt.figure(figsize=(6,6)) #Tamaño de la gráfica
sns.barplot(data=forbes_df.head(10), x='Country', y='Market Value (Billions)', palette='Set2', errorbar=None)
plt.title('Valor del mercado por País')
```

```python
plt.figure(figsize=(5,3)) #Tamaño de la gráfica
sns.boxplot(x='Year Founded', data=forbes_df)
plt.title('Año de fundación')
```

```python
sns.scatterplot(data=forbes_df, x='Year Founded', y='Revenue (Billions)', color='pink')
plt.title('Año de fundación vs Ingresos')
```

```python
sns.scatterplot(data=forbes_df, x='Profits (Billions)', y='Revenue (Billions)', color='purple')
plt.title('Beneficio vs Ingresos')
```

```python
sns.heatmap(round(forbes_df.corr(numeric_only=True),2),annot=True,cmap='Set2')
plt.title('Diagrama de calor')
```
